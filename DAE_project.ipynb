Converting post time to date time

```python
import pandas as pd

# Convert post_time to datetime
df['post_time'] = pd.to_datetime(df['post_time'], errors='coerce')

# Extract posting hour
df['hour'] = df['post_time'].dt.hour

print("Posting Hour:")
print(df[['post_time', 'hour']].head(10))
```

Delete rows

```python
df= df.dropna()
print(df)
```

For checking it is skew distribution or not

```python
df[["likes", "comments", "shares"]].skew()
```

It is a skew distribution, So I perform median operation

```python
df["likes"] = df["likes"].fillna(df["likes"].median())
df["comments"] = df["comments"].fillna(df["comments"].median())
df["shares"] = df["shares"].fillna(df["shares"].median())
print(df["likes"])
print(df["comments"])
print(df["shares"])
```

To remove duplicate records

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

Time-Series Analysis and Data Visualization

```python
import matplotlib.pyplot as plt

# Ensure 'post_time' is a datetime object
df['post_time'] = pd.to_datetime(df['post_time'])

# Set 'post_time' as the index for time-series plotting
df_time = df.set_index('post_time')

# Resample the data daily and calculate the mean for likes, comments, and shares
daily_engagement = df_time[['likes', 'comments', 'shares']].resample('D').mean().fillna(0)

# Create the line graph
plt.figure(figsize=(12, 6))
for column in daily_engagement.columns:
    plt.plot(daily_engagement.index, daily_engagement[column], marker='o', linestyle='-', label=column)

plt.title('Daily Trend of Social Media Engagement Metrics')
plt.xlabel('Date')
plt.ylabel('Average Count')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Pie chart

```python
total_likes = df['likes'].sum()
total_comments = df['comments'].sum()
total_shares = df['shares'].sum()

values = [total_likes, total_comments, total_shares]
labels = ['Likes', 'Comments', 'Shares']

plt.figure(figsize=(7,7))

plt.pie(values, labels=labels, autopct='%1.1f%%')

plt.title('Social Media Engagement Distribution')

plt.show()
```

```python
import pandas as pd

# Select engagement columns
columns = ['likes', 'comments', 'shares']

# Detect outliers using IQR method
for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Display outliers
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print("Outliers in", col)
    print(outliers)
```

Line Graph

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert post_time to datetime
df['post_time'] = pd.to_datetime(df['post_time'])

# Extract posting hour
df['hour'] = df['post_time'].dt.hour

# Calculate average engagement by hour
hour_engagement = df.groupby('hour')[['likes', 'comments', 'shares']].mean()

# Create line chart
hour_engagement.plot(kind='line', marker='o')

plt.title("Engagement by Posting Time")
plt.xlabel("Posting Hour")
plt.ylabel("Average Engagement")
plt.xticks(range(0, 24))
plt.grid()
plt.show()
```

Correlation and Outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select engagement columns
engagement = df[['likes', 'comments', 'shares']]

# Identify correlations
correlation = engagement.corr()
print("Correlation:")
print(correlation)

# Display correlation heatmap
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation between Engagement Metrics")
plt.show()

# Identify outliers using boxplot
sns.boxplot(data=engagement)
plt.title("Outliers in Engagement Data")
plt.xlabel("Engagement Metrics")
plt.ylabel("Values")
plt.show()
```

Bivariate Analysis: Post Day vs. Engagement Metrics

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Boxplot for 'likes' vs 'post_day'
plt.figure(figsize=(12, 7))
sns.boxplot(data=df, x='post_day', y='likes', palette='coolwarm')
plt.title('Likes by Post Day')
plt.xlabel('Post Day')
plt.ylabel('Likes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Boxplot for 'comments' vs 'post_day'
plt.figure(figsize=(12, 7))
sns.boxplot(data=df, x='post_day', y='comments', palette='coolwarm')
plt.title('Comments by Post Day')
plt.xlabel('Post Day')
plt.ylabel('Comments')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Boxplot for 'shares' vs 'post_day'
plt.figure(figsize=(12, 7))
sns.boxplot(data=df, x='post_day', y='shares', palette='coolwarm')
plt.title('Shares by Post Day')
plt.xlabel('Post Day')
plt.ylabel('Shares')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Multivariate Analysis: Numerical Features vs. Post Type

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a figure with subplots for each numerical feature
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle('Distribution of Engagement Metrics by Post Type', fontsize=16)

# Boxplot for 'likes' vs 'post_type'
sns.boxplot(data=df, x='post_type', y='likes', ax=axes[0], palette='viridis')
axes[0].set_title('Likes by Post Type')
axes[0].set_xlabel('Post Type')
axes[0].set_ylabel('Likes')
axes[0].tick_params(axis='x', rotation=45)

# Boxplot for 'comments' vs 'post_type'
sns.boxplot(data=df, x='post_type', y='comments', ax=axes[1], palette='viridis')
axes[1].set_title('Comments by Post Type')
axes[1].set_xlabel('Post Type')
axes[1].set_ylabel('Comments')
axes[1].tick_params(axis='x', rotation=45)

# Boxplot for 'shares' vs 'post_type'
sns.boxplot(data=df, x='post_type', y='shares', ax=axes[2], palette='viridis')
axes[2].set_title('Shares by Post Type')
axes[2].set_xlabel('Post Type')
axes[2].set_ylabel('Shares')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Adjust layout to prevent title overlap
plt.show()
```
